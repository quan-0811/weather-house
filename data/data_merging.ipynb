```python
import pandas as pd
import numpy as np
import os

# Define file paths (adjust folder path if necessary)
file1_path = 'reordered_data/merged_output.csv'
file2_path = 'reordered_data/merged_output1.csv'
output_path = 'final/final_data.csv'
```

```python
# 1. Read the CSV files
df1 = pd.read_csv(file1_path)
df2 = pd.read_csv(file2_path)

print(f"Records in File 1: {len(df1)}")
print(f"Records in File 2: {len(df2)}")

# 2. Handle location_id collision
# Find the max ID in the first file to calculate the offset
max_id_df1 = df1['location_id'].max()
offset = max_id_df1 + 1

print(f"Offsetting File 2 location_ids by: {offset}")
df2['location_id'] = df2['location_id'] + offset

# 3. Concatenate the dataframes
combined_df = pd.concat([df1, df2], ignore_index=True)

# 4. Sort by time, shuffle within same time
# Create a temporary random column for secondary sorting
combined_df['random_seed'] = np.random.rand(len(combined_df))

# Sort by 'time' first, then by the random seed
combined_df = combined_df.sort_values(by=['time', 'random_seed'])

# Remove the temporary column
combined_df = combined_df.drop(columns=['random_seed'])

# 5. Save the result
combined_df.to_csv(output_path, index=False)
print(f"Successfully saved {len(combined_df)} records to {output_path}")
```

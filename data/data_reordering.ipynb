```python
import os
import pandas as pd
import re
from pathlib import Path

# Define input and output directories
INPUT_FOLDER = 'data_bigdata'
OUTPUT_FOLDER = 'merged_data'

# Create output folder if it doesn't exist
os.makedirs(OUTPUT_FOLDER, exist_ok=True)
```

```python
def merge_sample_with_stations(sample_path, stations_path, output_path):
    """
    Reads sample data and station data, joins them on 'location_id', 
    and saves the result to the output path.
    """
    print(f"Reading files:\n - {sample_path}\n - {stations_path}")
    
    # Read CSVs
    df_sample = pd.read_csv(sample_path)
    df_stations = pd.read_csv(stations_path)
    
    # Merge on location_id
    # We use a left join to keep all weather records and attach station info
    merged_df = pd.merge(df_sample, df_stations, on='location_id', how='left')
    
    # Reorder columns: stations columns first, then remaining sample columns
    station_cols = list(df_stations.columns)
    sample_cols = [c for c in df_sample.columns if c not in station_cols]
    merged_df = merged_df[station_cols + sample_cols]
    
    # Save to CSV
    merged_df.to_csv(output_path, index=False)
    print(f"Successfully saved merged file to {output_path}")
    
    return merged_df
```

```python
stations_file = 'raw/stations1.csv' 
sample_file = 'raw/sample1.csv'
output_file = os.path.join(OUTPUT_FOLDER, 'merged_output1.csv')
df = merge_sample_with_stations(sample_file, stations_file, output_file)
```
